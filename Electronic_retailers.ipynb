```python
products["Unit Cost USD"] = (
    products["Unit Cost USD"]
    .str.replace("$", "", regex=False)
    .str.replace(",", "", regex=False)
    .str.strip()
    .astype(float)
)

products["Unit Price USD"] = (
    products["Unit Price USD"]
    .str.replace("$", "", regex=False)
    .str.replace(",", "", regex=False)
    .str.strip()
    .astype(float)
)
```

```python
from sqlalchemy import create_engine
from urllib.parse import quote_plus

username = "postgres"
password = quote_plus("Sanju@123")
host = "localhost"
port = "5432"
database = "Maven_Market_Analytics"

engine = create_engine(
    f"postgresql+psycopg2://{username}:{password}@{host}:{port}/{database}"
)

print("Connection object created successfully!")
```

```python
with engine.connect() as connection:
    print("Database connected successfully!")
```

```python
df.to_sql(
    "sales",
    con=engine,
    if_exists="replace",
    index=False
)

customers.to_sql(
    "customers",
    con=engine,
    if_exists="replace",
    index=False
)

stores.to_sql(
    "stores",
    con=engine,
    if_exists="replace",
    index=False
)

products.to_sql(
    "products",
    con=engine,
    if_exists="replace",
    index=False
)

exchange_rates.to_sql(
    "exchange_rates",
    con=engine,
    if_exists="replace",
    index=False
)

print("All tables loaded successfully!")
```
